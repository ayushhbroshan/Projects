```python
import pandas as pd
import numpy as np
df=pd.read_csv('https://raw.githubusercontent.com/ayushhbroshan/files/refs/heads/main/west_bengal_2026_complete_292.csv')
df
```

```python
# @title
df['Assets_Amount'] = (
    df['Assets']
      .str.split(r'~|\(', regex=True)
      .str[0]
      .str.replace('Rs', '', regex=False)
      .str.replace(',', '', regex=False)
      .str.strip()
      .astype(int)
)

df['Liabilities_Amount'] = (
    df['Liabilities']
      .str.split(r'~|\(', regex=True)
      .str[0]
      .str.replace('Rs', '', regex=False)
      .str.replace(',', '', regex=False)
      .str.strip()
      .astype(int)
)

df.drop(columns=['Assets', 'Liabilities'], inplace=True)
df
```

```python
count=df['Candidate'].value_counts()
candidates=count[count==2].index

constituency_2=df[df['Candidate'].isin(candidates)]
```

```python
df.loc[df['Sno'] == 32, ['Criminal Cases', 'Education']] = [3, 'Post Graduate',]
df.loc[df['Sno'] == 31, ['Assets_Amount', 'Liabilities_Amount']] = [39036176, 8409674,]
```

```python
df['Constituency'] = (
    df.groupby('Candidate')['Constituency']
    .transform(lambda x: ', '.join(x))
)

df = df.drop_duplicates(subset='Candidate')
```

```python
df["Education"] = df["Education"].replace(
    "Graduate Professional",
    "Graduate"
)

df["Education"].value_counts()
```

```python
lower_education = [
    '12th Pass',
    '10th Pass',
    '8th Pass',
    'Literate',
    '5th Pass',
    'Illiterate'
]

higher_education = [
    'Graduate',
    'Post Graduate',
    'Doctorate'
]

df["Education_level"] = "Others"

df.loc[
    df["Education"].isin(lower_education),
    "Education_level"
] = "Lower Education"

df.loc[
    df["Education"].isin(higher_education),
    "Education_level"
] = "Higher Education"

df["Education_level"].value_counts()
```

```python
#cleaned data
df
```

```python
df['net_worth']=df['Assets_Amount']-df['Liabilities_Amount']
df.sort_values('net_worth', ascending=False).head(20)
```

```python
# @title 1. Top 20 Candidates by Declared Net Worth (Interactive Dashboard)

import pandas as pd
import plotly.graph_objects as go

# ==========================================================
# DATA PREPARATION
# ==========================================================

df["net_worth"] = df["Assets_Amount"] - df["Liabilities_Amount"]

top20 = (
    df.sort_values("net_worth", ascending=False)
      .head(20)
      .copy()
)

top20["NetWorthCr"] = (top20["net_worth"] / 1e7).round(2)

# ==========================================================
# PARTY COLOURS
# ==========================================================

party_colors = {
    "BJP": "#FF9933",
    "AITC": "#009E60",
    "INC": "#1976D2",
    "CPI(M)": "#D32F2F",
    "All India Secular Front": "#7B1FA2",
    "Aam Janata Unnayan party": "#795548"
}

top20["Color"] = (
    top20["Party"]
    .map(party_colors)
    .fillna("#90A4AE")
)

# ==========================================================
# FIGURE
# ==========================================================

fig = go.Figure()

# ----------------------------------------------------------
# Main Bar Chart
# ----------------------------------------------------------

fig.add_trace(

    go.Bar(

        x=top20["Candidate"],

        y=top20["NetWorthCr"],

        marker=dict(

            color=top20["Color"],

            line=dict(
                color="black",
                width=1
            )
        ),

        text=[f"₹{v:.1f} Cr" for v in top20["NetWorthCr"]],

        textposition="outside",

        customdata=top20[
            [
                "Party",
                "Constituency",
                "Education",
                "Assets_Amount",
                "Liabilities_Amount"
            ]
        ],

        hovertemplate=
        "<b>%{x}</b><br><br>"
        "<b>Party:</b> %{customdata[0]}<br>"
        "<b>Constituency:</b> %{customdata[1]}<br>"
        "<b>Education:</b> %{customdata[2]}<br><br>"
        "<b>Declared Assets:</b> ₹%{customdata[3]:,.0f}<br>"
        "<b>Declared Liabilities:</b> ₹%{customdata[4]:,.0f}<br>"
        "<b>Net Worth:</b> ₹%{y:.2f} Cr"
        "<extra></extra>",

        showlegend=False
    )
)

# ==========================================================
# PARTY LEGEND
# ==========================================================

for party, color in party_colors.items():

    fig.add_trace(

        go.Scatter(

            x=[None],
            y=[None],

            mode="markers",

            marker=dict(
                size=12,
                color=color
            ),

            name=party
        )
    )

# ==========================================================
# TOP 5 PANEL
# ==========================================================

top5 = top20.head(5)

summary = "<b>🏆 TOP 5 CANDIDATES</b><br><br>"

for i, (_, row) in enumerate(top5.iterrows(), 1):

    summary += (
        f"<span style='color:{party_colors.get(row['Party'],'black')}'>●</span> "
        f"<b>#{i} {row['Candidate']}</b><br>"
        f"&nbsp;&nbsp;{row['Party']}<br>"
        f"&nbsp;&nbsp;{row['Constituency']}<br>"
        f"&nbsp;&nbsp;{row['Education']}<br><br>"
    )

# ==========================================================
# LAYOUT
# ==========================================================

fig.update_layout(

    template="plotly_white",

    width=1500,

    height=850,

    title=dict(

        text=(
            "<b>WEST BENGAL ASSEMBLY ELECTION 2026</b><br>"
            "<span style='font-size:20px;'>"
            "Top 20 Candidates by Declared Net Worth"
            "</span>"
        ),

        x=0.5,

        font=dict(
            family="Arial",
            size=28,
            color="#0D1B2A"
        )
    ),

    xaxis=dict(

        title="",

        tickangle=-65,

        tickfont=dict(size=10)

    ),

    yaxis=dict(

        title="<b>Declared Net Worth (₹ Crore)</b>",

        gridcolor="rgba(0,0,0,0.08)",

        zeroline=False

    ),

    paper_bgcolor="white",

    plot_bgcolor="white",

    legend=dict(

        title="<b>Political Party</b>",

        orientation="h",

        y=.95,

        x=0.5,

        xanchor="center",

        font=dict(size=11)

    ),

    margin=dict(
        l=70,
        r=320,
        t=120,
        b=170
    )

)

# ==========================================================
# TOP 5 BOX
# ==========================================================

fig.add_annotation(

    x=1.02,
    y=1.05,

    xref="paper",
    yref="paper",

    text=summary,

    showarrow=False,

    align="left",

    bordercolor="#B0BEC5",

    borderwidth=1,

    borderpad=8,

    bgcolor="#F8F9FA",

    font=dict(size=11)

)

# ==========================================================
# SOURCE
# ==========================================================

fig.add_annotation(

    x=0,
    y=-0.18,

    xref="paper",
    yref="paper",

    text=(
        "<i>Source: MyNeta (ADR & National Election Watch) • "
        "Net Worth = Declared Assets − Declared Liabilities</i>"
    ),

    showarrow=False,

    font=dict(
        size=11,
        color="gray"
    )

)

fig.show()
```

```python

# @title 2. What is the education distribution among candidates?
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px

# ==========================================================
# EDUCATION DISTRIBUTION
# ==========================================================

education_distribution = (
    df["Education"]
    .value_counts()
    .reset_index()
)

education_distribution.columns = ["Education", "Count"]

education_distribution["Percentage"] = (
    education_distribution["Count"]
    / education_distribution["Count"].sum()
    * 100
).round(1)

total_candidates = education_distribution["Count"].sum()

# ==========================================================
# BLUE GRADIENT (DARKER = MORE CANDIDATES)
# ==========================================================

colors = px.colors.sequential.Blues[::-1]

if len(education_distribution) <= len(colors):
    pie_colors = colors[:len(education_distribution)]
else:
    pie_colors = px.colors.sample_colorscale(
        "Blues",
        [i/(len(education_distribution)-1)
         for i in range(len(education_distribution))]
    )

# ==========================================================
# LABELS
# Show percentage only if >= 5%
# ==========================================================

labels = [
    f"{p:.1f}%"
    if p >= 5
    else ""
    for p in education_distribution["Percentage"]
]

# ==========================================================
# DONUT CHART
# ==========================================================

fig = go.Figure()

fig.add_trace(

    go.Pie(

        labels=education_distribution["Education"],

        values=education_distribution["Count"],

        hole=0.60,

        marker=dict(
            colors=pie_colors,
            line=dict(color="white", width=2)
        ),

        text=labels,

        textinfo="text",

        textfont=dict(
            size=15,
            color="black"
        ),

        hovertemplate=
        "<b>%{label}</b><br><br>"
        "Candidates : <b>%{value}</b><br>"
        "Percentage : <b>%{percent}</b>"
        "<extra></extra>",

        sort=False
    )
)

# ==========================================================
# CENTER TEXT
# ==========================================================

fig.add_annotation(

    x=0.5,
    y=0.5,

    text=
    f"<b>{total_candidates}</b><br>"
    "<span style='font-size:18px'>Candidates</span>",

    showarrow=False,

    font=dict(
        size=30,
        color="#0D47A1"
    )
)

# ==========================================================
# TITLE
# ==========================================================

fig.update_layout(

   title=dict(
    text=(
        "<b>WEST BENGAL ASSEMBLY ELECTION 2026</b><br>"
        "<span style='font-size:20px;'>Educational Profile of Elected Members of the Legislative Assembly (MLAs)</span>"
    ),
    x=0.5,
    xanchor="center",
    font=dict(
        family="Arial",
        size=28,
        color="#0D1B2A"
    )
),

    legend=dict(

        title="<b>Education Level</b>",

        font=dict(size=13),

        x=1.02,

        y=0.5,

        bgcolor="rgba(255,255,255,0.9)"
    ),

    width=1000,

    height=700,

    paper_bgcolor="white",

    plot_bgcolor="white",

    margin=dict(
        l=20,
        r=150,
        t=90,
        b=30
    )
)

fig.show()
```

```python
# @title 3. Average assets and liabilities of competitive parties.
import plotly.graph_objects as go
import pandas as pd

# ==========================================================
# DATA PREPARATION
# ==========================================================

# Keep only parties with >=5 seats
party_counts = df["Party"].value_counts()
major_parties = party_counts[party_counts >= 5].index

filtered = df[df["Party"].isin(major_parties)]

summary = (
    filtered.groupby("Party")[["Assets_Amount","Liabilities_Amount"]]
    .mean()/1e7
).round(2)

summary = summary.reset_index()

# ==========================================================
# PARTY COLORS
# ==========================================================

party_colors = {
    "BJP":"#FF9933",
    "AITC":"#00A651",
    "INC":"#1976D2",
    "CPI(M)":"#D32F2F",
    "All India Secular Front":"#7B1FA2",
    "Aam Janata Unnayan party":"#795548"
}

axis_colors = [
    party_colors.get(p,"black")
    for p in summary["Party"]
]

# ==========================================================
# FIGURE
# ==========================================================

fig = go.Figure()

# Assets
fig.add_trace(
    go.Bar(
        x=summary["Party"],
        y=summary["Assets_Amount"],
        name="Average Assets",
        marker_color="#1565C0",
        text=[f"₹{x:.1f} Cr" for x in summary["Assets_Amount"]],
        textposition="outside",
        hovertemplate=
        "<b>%{x}</b><br>"
        "Average Assets: ₹%{y:.2f} Cr"
        "<extra></extra>"
    )
)

# Liabilities
fig.add_trace(
    go.Bar(
        x=summary["Party"],
        y=summary["Liabilities_Amount"],
        name="Average Liabilities",
        marker_color="#D32F2F",
        text=[f"₹{x:.1f} Cr" for x in summary["Liabilities_Amount"]],
        textposition="outside",
        hovertemplate=
        "<b>%{x}</b><br>"
        "Average Liabilities: ₹%{y:.2f} Cr"
        "<extra></extra>"
    )
)

# ==========================================================
# SEAT DISTRIBUTION BOX
# ==========================================================

seat_text = "<b>Party Seat Distribution</b><br><br>"

for party,seats in party_counts.items():

    color = party_colors.get(party,"black")

    seat_text += (
        f"<span style='color:{color}'>●</span> "
        f"<b>{party}</b> : {seats}<br>"
    )

# ==========================================================
# LAYOUT
# ==========================================================

fig.update_layout(

    template="plotly_white",

    barmode="group",

    width=1200,

    height=700,

    title=dict(
        text=(
            "<b>WEST BENGAL ASSEMBLY ELECTION 2026</b><br>"
            "<sup>Party-wise Average Declared Assets & Liabilities of Elected MLAs</sup>"
        ),
        x=0.5,
        font=dict(size=24)
    ),

    yaxis_title="<b>Average Value (₹ Crore)</b>",

    xaxis_title="",

    font=dict(
        family="Arial",
        size=14
    ),

    legend=dict(
        orientation="h",
        y=1.08,
        x=0.5,
        xanchor="center"
    ),

    margin=dict(
        l=60,
        r=250,
        t=110,
        b=70
    )
)

# ==========================================================
# PARTY COLORED X LABELS
# ==========================================================

fig.update_xaxes(
    tickfont=dict(size=13)
)

# ==========================================================
# SIDE PANEL
# ==========================================================

fig.add_annotation(

    x=1.18,
    y=0.98,

    xref="paper",
    yref="paper",

    align="left",

    showarrow=False,

    bordercolor="black",

    borderwidth=1,

    borderpad=8,

    bgcolor="#F8F9FA",

    text=seat_text
)

# ==========================================================
# FOOTNOTE
# ==========================================================

fig.add_annotation(

    x=0,
    y=-0.18,

    xref="paper",
    yref="paper",

    showarrow=False,

    align="left",

    text=(
        "<i>Source: MyNeta (ADR & National Election Watch). "
        "Values represent average declared assets and liabilities of elected candidates.</i>"
    )
)

fig.show()
```

```python
# @title 4. Maximum Declared Assets vs Maximum Declared Liabilities by Political Party

import plotly.graph_objects as go

# ==========================================================
# DATA
# ==========================================================

summary = (
    df.groupby("Party")[["Assets_Amount", "Liabilities_Amount"]]
      .max()
      .reset_index()
)

summary["AssetsCr"] = (summary["Assets_Amount"] / 1e7).round(2)
summary["LiabilitiesCr"] = (summary["Liabilities_Amount"] / 1e7).round(2)

summary = summary.sort_values("AssetsCr", ascending=True)

# ==========================================================
# FIGURE
# ==========================================================

fig = go.Figure()

# ----------------------------------------------------------
# CONNECTING LINES
# ----------------------------------------------------------

for _, row in summary.iterrows():

    fig.add_trace(
        go.Scatter(
            x=[row["LiabilitiesCr"], row["AssetsCr"]],
            y=[row["Party"], row["Party"]],
            mode="lines",
            line=dict(color="#BDBDBD", width=4),
            hoverinfo="skip",
            showlegend=False
        )
    )

# ----------------------------------------------------------
# ASSET MARKERS
# ----------------------------------------------------------

fig.add_trace(

    go.Scatter(

        x=summary["AssetsCr"],

        y=summary["Party"],

        mode="markers",

        name="Maximum Assets",

        marker=dict(
            size=18,
            color="#1565C0",
            line=dict(color="white", width=2)
        ),

        hovertemplate=
        "<b>%{y}</b><br>"
        "Maximum Declared Assets<br>"
        "₹%{x:.2f} Cr"
        "<extra></extra>"
    )
)

# ----------------------------------------------------------
# LIABILITY MARKERS
# ----------------------------------------------------------

fig.add_trace(

    go.Scatter(

        x=summary["LiabilitiesCr"],

        y=summary["Party"],

        mode="markers",

        name="Maximum Liabilities",

        marker=dict(
            size=18,
            color="#D32F2F",
            line=dict(color="white", width=2)
        ),

        hovertemplate=
        "<b>%{y}</b><br>"
        "Maximum Declared Liabilities<br>"
        "₹%{x:.2f} Cr"
        "<extra></extra>"
    )
)


# ==========================================================
# SMART LABELS (CORRECTED FOR ANY ASSET/LIABILITY ORDER)
# ==========================================================

for _, row in summary.iterrows():
    asset = row["AssetsCr"]
    liability = row["LiabilitiesCr"]
    gap = abs(asset - liability)

    # 1. VERY CLOSE VALUES -> Stack vertically
    if gap < 0.5:
        # Higher value goes slightly up, lower value goes slightly down
        asset_yshift = 16 if asset >= liability else -16
        liab_yshift = -16 if asset >= liability else 16

        fig.add_annotation(
            x=asset,
            y=row["Party"],
            text=f"<b>  ₹{asset:.1f} Cr  </b>",
            showarrow=False,
            yshift=asset_yshift,
            font=dict(color="#1565C0", size=12)
        )

        fig.add_annotation(
            x=liability,
            y=row["Party"],
            text=f"<b>  ₹{liability:.1f} Cr  </b>",
            showarrow=False,
            yshift=liab_yshift,
            font=dict(color="#D32F2F", size=12)
        )

    # 2. NORMAL / FAR APART VALUES -> Dynamic Left/Right shift
    else:
        # Asset label position
        asset_xshift = 50 if asset > liability else -50
        fig.add_annotation(
            x=asset,
            y=row["Party"],
            text=f"<b>  ₹{asset:.1f} Cr  </b>",
            showarrow=False,
            xshift=asset_xshift,
            font=dict(color="#1565C0", size=12)
        )

        # Liability label position
        liab_xshift = -50 if asset > liability else 50
        fig.add_annotation(
            x=liability,
            y=row["Party"],
            text=f"<b>  ₹{liability:.1f} Cr  </b>",
            showarrow=False,
            xshift=liab_xshift,
            font=dict(color="#D32F2F", size=12)
        )


# ==========================================================
# LAYOUT ADJUSTMENTS (Fixes Legend Overlap)
# ==========================================================

fig.update_layout(
    template="plotly_white",
    width=1200,
    height=700,
    title=dict(
        text=(
            "<b>WEST BENGAL ASSEMBLY ELECTION 2026</b><br>"
            "<span style='font-size:18px; color:#555555'>"
            "Maximum Declared Assets vs Maximum Declared Liabilities by Political Party"
            "</span>"
        ),
        x=0.5,
        y=0.96, # Moved title slightly down
        font=dict(size=24, color="#0D1B2A")
    ),
    xaxis=dict(
        title="<b>Amount (₹ Crore)</b>",
        showgrid=True,
        gridcolor="rgba(0,0,0,0.08)",
        zeroline=False
    ),
    yaxis=dict(
        title="",
        categoryorder="array",
        categoryarray=summary["Party"]
    ),
    legend=dict(
        orientation="h",
        x=0.5,
        xanchor="center",
        y=1.02 # Adjusted legend so it doesn't collide with title
    ),
    paper_bgcolor="white",
    plot_bgcolor="white",
    margin=dict(
        l=150, # Slightly extra room for long party names
        r=80,
        t=140, # Extra top margin for title & legend
        b=60
    )
)

fig.show()
```

### ⚠️ Notable Financial Flags

The most striking case is **AITC**, where maximum declared liabilities (**₹43.7 Cr**) exceed maximum declared assets (**₹39.5 Cr**) by approximately **₹4.2 Cr**. This produces an unusual liability-to-asset position and warrants verification of the underlying affidavit and the **composition and nature of those liabilities**.

However, this should **not automatically be interpreted as financial loss, tax avoidance, or wrongdoing**. The chart only identifies an unusual financial position; further investigation of the original affidavit would be required to understand it.

### 📌 Analytical Takeaway

### ⚠️ Political Context & Possible Implication

This finding becomes more noteworthy in the context of the **2026 election**, where AITC entered the election as the incumbent governing party and BJP emerged with the largest number of winners in this dataset. One possible line of investigation is whether BJP's candidate selection was associated with a **higher prevalence of candidates with declared criminal cases**, or whether differences arise from the political environment, constituency characteristics, opposition dynamics, or the types of cases involved.

However, the chart **cannot establish why these differences exist**. In particular, it would be inappropriate to conclude that BJP candidates are inherently more "prone" to criminality based solely on declared cases. ### 📌 Political Consulting Takeaway

```python
# @title 10. Criminal case profile by party and education level
import plotly.graph_objects as go

# ==========================================================
# Prepare Data
# ==========================================================

summary = (
    df.groupby(["Education_level", "Party"])["Criminal Cases"]
      .agg(
          Candidates="count",
          Average_Cases="mean",
          Total_Cases="sum"
      )
      .round({"Average_Cases": 2})
      .reset_index()
)

# ==========================================================
# Pivot Tables
# ==========================================================

avg_cases = summary.pivot(
    index="Education_level",
    columns="Party",
    values="Average_Cases"
)

candidate_count = summary.pivot(
    index="Education_level",
    columns="Party",
    values="Candidates"
)

total_cases = summary.pivot(
    index="Education_level",
    columns="Party",
    values="Total_Cases"
)

# ==========================================================
# Annotation Text
# ==========================================================

annotation_text = []

for i in range(len(avg_cases.index)):

    row = []

    for j in range(len(avg_cases.columns)):

        c = candidate_count.iloc[i, j]
        t = total_cases.iloc[i, j]

        if pd.isna(c):
            row.append("")
        else:
            row.append(f"{int(c)} | {int(t)}")

    annotation_text.append(row)

# ==========================================================
# Heatmap
# ==========================================================

fig = go.Figure(

    data=go.Heatmap(

        z=avg_cases.values,

        x=avg_cases.columns,

        y=avg_cases.index,

        colorscale="Reds",

        colorbar=dict(
            title="Average Criminal Cases"
        ),

        text=annotation_text,

        texttemplate="%{text}",

        textfont=dict(
            color="black",
            size=12
        ),

        hovertemplate=
        "<b>Party:</b> %{x}<br>" +
        "<b>Education:</b> %{y}<br>" +
        "<b>Average Criminal Cases:</b> %{z:.2f}<br>" +
        "<b>Candidates | Total Cases:</b> %{text}" +
        "<extra></extra>"

    )

)

# ==========================================================
# Layout
# ==========================================================

fig.update_layout(

    title=dict(
        text="<b>Criminal Case Profile by Party and Education Level</b>",
        x=0.5,
        font=dict(size=24)
    ),

    xaxis_title="Political Party",

    yaxis_title="Education Level",

    paper_bgcolor="white",

    plot_bgcolor="white",

    width=1100,

    height=450,

    font=dict(
        family="Arial",
        size=13
    )
)

fig.show()
```

```python
# @title
education_rank = {
    "Illiterate": 0,
    "Literate": 1,
    "5th Pass": 2,
    "8th Pass": 3,
    "10th Pass": 4,
    "12th Pass": 5,
    "Graduate": 6,
    "Post Graduate": 7,
    "Doctorate": 8
}

df["Education_Rank"] = df["Education"].map(education_rank)

import plotly.graph_objects as go
import numpy as np
from scipy.stats import spearmanr

# ==========================================================
# EDUCATION ORDER
# ==========================================================

education_order = [
    "Illiterate",
    "Literate",
    "5th Pass",
    "8th Pass",
    "10th Pass",
    "12th Pass",
    "Graduate",
    "Post Graduate",
    "Doctorate"
]

education_rank = {
    education: i
    for i, education in enumerate(education_order)
}

# ==========================================================
# PREPARE DATA
# ==========================================================

analysis_df = df[
    [
        "Education",
        "Education_Rank",
        "Assets_Amount",
        "Candidate",
        "Party",
        "Constituency"
    ]
].copy()

analysis_df = analysis_df.dropna(
    subset=["Education_Rank", "Assets_Amount"]
)

analysis_df = analysis_df[
    analysis_df["Assets_Amount"] > 0
]

# ==========================================================
# SPEARMAN CORRELATION
# ==========================================================

corr, p_value = spearmanr(
    analysis_df["Education_Rank"],
    analysis_df["Assets_Amount"]
)

# ==========================================================
# JITTER EDUCATION RANK
# ==========================================================

np.random.seed(42)

analysis_df["Education_Jitter"] = (
    analysis_df["Education_Rank"]
    + np.random.uniform(
        -0.16,
        0.16,
        len(analysis_df)
    )
)

# ==========================================================
# INTERPRETATION
# ==========================================================

if abs(corr) < 0.10:
    relationship = "Very Weak / No Monotonic Relationship"

elif abs(corr) < 0.30:
    relationship = "Weak Relationship"

elif abs(corr) < 0.50:
    relationship = "Moderate Relationship"

elif abs(corr) < 0.70:
    relationship = "Strong Relationship"

else:
    relationship = "Very Strong Relationship"


significance = (
    "Statistically Significant"
    if p_value < 0.05
    else "Not Statistically Significant"
)

# ==========================================================
# FIGURE
# ==========================================================

fig = go.Figure()

# ==========================================================
# CANDIDATE POINTS
# ==========================================================

fig.add_trace(

    go.Scatter(

        x=analysis_df["Education_Jitter"],

        y=analysis_df["Assets_Amount"],

        mode="markers",

        name="Winning Candidates",

        marker=dict(

            size=9,

            color="#263238",

            opacity=0.65,

            line=dict(
                color="white",
                width=1
            )

        ),

        customdata=analysis_df[
            [
                "Education",
                "Candidate",
                "Party",
                "Constituency"
            ]
        ],

        hovertemplate=

        "<b>%{customdata[1]}</b><br>"
        "Education: %{customdata[0]}<br>"
        "Party: %{customdata[2]}<br>"
        "Constituency: %{customdata[3]}<br>"
        "Assets: ₹ %{y:,.0f}"
        "<extra></extra>"

    )

)

# ==========================================================
# MEAN ASSETS BY EDUCATION
# ==========================================================

mean_assets = (
    analysis_df
    .groupby("Education_Rank")["Assets_Amount"]
    .mean()
    .reindex(range(len(education_order)))
)

# ==========================================================
# MEAN TREND LINE
# ==========================================================

fig.add_trace(

    go.Scatter(

        x=list(range(len(education_order))),

        y=mean_assets.values,

        mode="lines+markers",

        name="Mean Assets",

        line=dict(
            color="#8B0000",
            width=4
        ),

        marker=dict(

            size=11,

            color="#8B0000",

            line=dict(
                color="white",
                width=2
            )

        ),

        hovertemplate=

        "<b>%{text}</b><br>"
        "Mean Assets: ₹ %{y:,.0f}"
        "<extra></extra>",

        text=education_order

    )

)

# ==========================================================
# LAYOUT
# ==========================================================

fig.update_layout(

    title=dict(

        text=(
            "<b>Educational Attainment vs Candidate Wealth</b>"
            "<br>"
            "<span style='font-size:15px;color:#555'>"
            "Distribution of Assets Across Educational Qualifications"
            "</span>"
        ),

        x=0.5,

        font=dict(
            size=25,
            color="#1F1F1F"
        )

    ),

    xaxis=dict(

        title="<b>Educational Qualification</b>",

        tickmode="array",

        tickvals=list(range(len(education_order))),

        ticktext=education_order,

        tickangle=-35,

        range=[
            -0.5,
            len(education_order) - 0.5
        ],

        showgrid=False,

        showline=True,

        linecolor="#555555",

        linewidth=1.5

    ),

    yaxis=dict(

        title="<b>Assets (₹) — Log Scale</b>",

        type="log",

        gridcolor="#E5E5E5",

        gridwidth=1,

        zeroline=False

    ),

    paper_bgcolor="white",

    plot_bgcolor="#FAFAFA",

    font=dict(

        family="Arial",

        size=13,

        color="#222222"

    ),

    width=1350,

    height=720,

    margin=dict(

        l=90,

        r=300,

        t=120,

        b=130

    ),

    legend=dict(

        orientation="h",

        x=0.5,

        xanchor="center",

        y=1.02

    )

)

# ==========================================================
# STATISTICAL INSIGHT PANEL
# ==========================================================

insight_text = (

    "<b>RELATIONSHIP SUMMARY</b><br><br>"

    f"<span style='font-size:22px;color:#8B0000'>"
    f"<b>ρ = {corr:.3f}</b>"
    f"</span><br>"

    f"<b>Spearman Correlation</b><br><br>"

    f"<b>P-value:</b> {p_value:.4f}<br><br>"

    f"<b>{relationship}</b><br>"

    f"<span style='color:#555'>"
    f"{significance}"
    f"</span><br><br>"

    "● <b>Each dot</b> = winning candidate<br>"
    "● <b>Dark red line</b> = mean assets<br>"
    "● <b>Log scale</b> = accommodates large wealth differences"

)

fig.add_annotation(

    x=1.30,

    y=0.05,

    xref="paper",

    yref="paper",

    text=insight_text,

    showarrow=False,

    align="left",

    valign="top",

    bordercolor="#8B0000",

    borderwidth=1.5,

    borderpad=14,

    bgcolor="#F7F7F7",

    font=dict(

        family="Arial",

        size=13,

        color="#222222"

    )

)

# ==========================================================
# SHOW
# ==========================================================

fig.show()

# ==========================================================
# PRINT STATISTICS
# ==========================================================

print(
    "Spearman Correlation:",
    round(corr, 3)
)

print(
    "P-value:",
    round(p_value, 4)
)
```
