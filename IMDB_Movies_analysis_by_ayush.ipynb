```python
# @title
import os
import json
import re
import numpy as np
import pandas as pd
from tqdm.auto import tqdm

df = pd.read_csv(os.path.join(path, "movies.csv"))

df = df.dropna()

df.head()
```

### 2. Data Cleaning

#### 2.1) Cleaning name,duration, votes column

```python
# @title
#name
df['name'] = df['name'].str.strip()
#duration
df['duration']=df['duration'].str.replace("min","").str.replace(",","")
#votes
df['votes'] = (
    df['votes']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('.0', '', regex=False)
)
```

#### 2.2) Cleaning year Column

```python
# @title
start_year = []
end_year   = []

for text in tqdm(df['year']):

    result = re.findall(r"\d{4}", text)

    if len(result) == 0:
        start_year.append('0')
        end_year.append('0')

    elif len(result) == 1:
        start_year.append(result[0])
        end_year.append(result[0])

    elif len(result) == 2:
        start_year.append(result[0])
        end_year.append(result[1])

df['start_year'] = start_year
df['start_year'] = df['start_year'].astype(int)

df['end_year']   = end_year
df['end_year'] = df['end_year'].astype(int)
```

#### 2.3) clean gross income

```python
# @title
df['gross_income'].value_counts()
df['gross_income']=df['gross_income'].str.replace(",","").str.replace("$","")
df['gross_income'] = (
    df['gross_income']
    .fillna('0')
    .str.upper()
    .str.replace('M', 'e6', regex=False)
    .str.replace('K', 'e3', regex=False)
    .astype(float)
)
df['gross_income']
```

#### 2.4) clean directors_id ,directors_name collumn

```python
# @title
df['directors_id'] = (
    df['directors_id']
    .str.findall(r'nm\d+')
    .str.join(',')
)
df['directors_name'] = df['directors_name'].str.replace('nm0000000', 'Anonymous')
d_id_list=[]
d_name_list=[]
count=0
for d_id , d_name in df[['directors_id','directors_name']].astype(str).values:
    if len(d_id.split(','))!=len(d_name.split(',')):
        d_id_list.append(np.nan)
        d_name_list.append(np.nan)
    else:
        d_id_list.append(d_id)
        d_name_list.append(d_name)

df['d_id']=d_id_list
df['d_name']=d_name_list

df=df.dropna()
directors_dict=dict(zip(d_id_list,d_name_list))

with open('directors.json','w') as f:
    json.dump(directors_dict,f)

del d_id_list
del d_name_list
del df['directors_name']
```

#### 2.5) Cleaning stars_id and stars_name

```python
# @title
df['stars_id']=df['stars_id'].str.replace('Anonymous','nm0000000')
df['stars_name']=df['stars_name'].str.replace('nm0000000','Anonymous')


star_id_list=[]
star_name_list=[]

for stars_id , stars_name in df[['stars_id','stars_name']].astype(str).values:
    if len(stars_id.split(','))!=len(stars_name.split(',')):
        star_id_list.append(np.nan)
        star_name_list.append(np.nan)
    else:
        star_id_list.append(stars_id)
        star_name_list.append(stars_name)

df['stars_id']=star_id_list
df['stars_name']=star_name_list

df=df.dropna()

stars_dict=dict(zip(star_id_list,star_name_list))

with open('stars.json','w') as f:
    json.dump(stars_dict,f)

del star_id_list
del star_name_list
del df['stars_name']
```

#### 3.3 Are there periods showing growth or decline in movie/series releases?

```python
# @title
import pandas as pd
import plotly.express as px

# ---------------------------------------------------
# Data Preparation
# ---------------------------------------------------

filtered_df = df[
    (df['start_year'] > 1900) &
    (df['start_year'] <= 2020)
]

movies_per_year = (
    filtered_df['start_year']
    .value_counts()
    .sort_index()
    .reset_index()
)

movies_per_year.columns = ['Year', 'Number of Movies']

# ---------------------------------------------------
# Calculate Growth / Decline
# ---------------------------------------------------

movies_per_year['Yearly Change'] = (
    movies_per_year['Number of Movies']
    .diff()
)

# Highest growth year
growth_row = movies_per_year.loc[
    movies_per_year['Yearly Change'].idxmax()
]

growth_year = int(growth_row['Year'])
growth_value = int(growth_row['Yearly Change'])

# Highest decline year
decline_row = movies_per_year.loc[
    movies_per_year['Yearly Change'].idxmin()
]

decline_year = int(decline_row['Year'])
decline_value = int(decline_row['Yearly Change'])

# Peak production year
peak_row = movies_per_year.loc[
    movies_per_year['Number of Movies'].idxmax()
]

peak_year = int(peak_row['Year'])
peak_movies = int(peak_row['Number of Movies'])

# ---------------------------------------------------
# Create Interactive Visualization
# ---------------------------------------------------

fig = px.line(
    movies_per_year,
    x='Year',
    y='Number of Movies',

    title="<b>IMDb</b> | Growth & Decline in Movie/Series Releases",

    markers=True
)

# ---------------------------------------------------
# IMDb-inspired Styling
# ---------------------------------------------------

fig.update_layout(
    template="plotly_dark",

    title_font=dict(
        size=24,
        color="#F5C518"
    ),

    xaxis_title="Year",
    yaxis_title="Number of Movies/Series Released",

    hovermode="x unified",

    width=1250,
    height=700,

    margin=dict(
        l=80,
        r=500,
        t=80,
        b=100
    )

)

# IMDb yellow styling
fig.update_traces(
    line=dict(
        color="#F5C518",
        width=3
    ),

    marker=dict(
        color="#F5C518",
        size=6
    )
)

# ---------------------------------------------------
# Highlight Growth Year
# ---------------------------------------------------

fig.add_scatter(
    x=[growth_year],
    y=[growth_row['Number of Movies']],

    mode='markers+text',

    text=[f"Highest Growth ({growth_year})"],

    textposition="top center",

    marker=dict(
        size=14,
        color="lime"
    ),

    name="Highest Growth"
)

# ---------------------------------------------------
# Highlight Decline Year
# ---------------------------------------------------

fig.add_scatter(
    x=[decline_year],
    y=[decline_row['Number of Movies']],

    mode='markers+text',

    text=[f"Largest Decline ({decline_year})"],

    textposition="bottom center",

    marker=dict(
        size=14,
        color="red"
    ),

    name="Largest Decline"
)

# ---------------------------------------------------
# Insight Annotation Box
# ---------------------------------------------------

insight_text = (
    f"<b>Key Insights</b><br>"
    f"• Peak production: {peak_movies} releases in {peak_year}<br>"
    f"• Highest yearly growth: +{growth_value} releases ({growth_year})<br>"
    f"• Largest yearly decline: {decline_value} releases ({decline_year})<br>"
    f"• Long-term production trends are clearly visible"
)

fig.add_annotation(
    xref="paper",
    yref="paper",

    x=0.5,
    y=0.5,

    showarrow=False,

    align="left",

    bgcolor="rgba(0,0,0,0.7)",

    bordercolor="#F5C518",
    borderwidth=1,

    font=dict(size=12),

    text=insight_text
)

# ---------------------------------------------------
# Show Visualization
# ---------------------------------------------------

fig.show()
```

```python
# @title
import pandas as pd
import plotly.express as px

# ---------------------------------------------------
# Data Preparation
# ---------------------------------------------------

# Convert rating column to numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Keep valid ratings and years between 1900–2020
filtered_df = df[
    (df['rating'] <= 10) &
    (df['end_year'] >= 1900) &
    (df['end_year'] <= 2020)
]

# Calculate average rating per year
result = (
    filtered_df
    .groupby('end_year')['rating']
    .mean()
    .reset_index(name='average_rating')
)

# Rename columns for cleaner plotting
result.columns = ['Year', 'Average Rating']

# ---------------------------------------------------
# Calculate Insights
# ---------------------------------------------------

highest_row = result.loc[result['Average Rating'].idxmax()]
lowest_row = result.loc[result['Average Rating'].idxmin()]

highest_year = int(highest_row['Year'])
highest_rating = round(highest_row['Average Rating'], 2)

lowest_year = int(lowest_row['Year'])
lowest_rating = round(lowest_row['Average Rating'], 2)

overall_avg_rating = round(result['Average Rating'].mean(), 2)

first_year = int(result['Year'].min())
last_year = int(result['Year'].max())

# ---------------------------------------------------
# Create Interactive Visualization
# ---------------------------------------------------

fig = px.line(
    result,
    x='Year',
    y='Average Rating',
    title="<b>IMDb</b> | Average Movie/Series Ratings Over Time",
    markers=True
)

# IMDb-inspired styling
fig.update_layout(
    template="plotly_dark",

    title_font=dict(
        size=24,
        color="#F5C518"
    ),

    xaxis_title="Year",
    yaxis_title="Average IMDb Rating",

    hovermode="x unified",

    margin=dict(r=260),

    xaxis=dict(
        range=[1900, 2020]
    )
)

# IMDb yellow styling
fig.update_traces(
    line=dict(
        color="#F5C518",
        width=3
    ),

    marker=dict(
        color="#F5C518",
        size=6
    )
)

# ---------------------------------------------------
# Add Insight Annotation Box
# ---------------------------------------------------

insight_text = (
    f"<b>Key Insights</b><br>"
    f"• Data range: {first_year} - {last_year}<br>"
    f"• Highest avg rating: {highest_rating} ({highest_year})<br>"
    f"• Lowest avg rating: {lowest_rating} ({lowest_year})<br>"
    f"• Overall avg IMDb rating: {overall_avg_rating}"
)

fig.add_annotation(
    xref="paper",
    yref="paper",

    x=1.18,   # move further right
    y=0.5,    # vertically centered

    showarrow=False,

    align="left",

    bgcolor="rgba(0,0,0,0.7)",

    bordercolor="#F5C518",
    borderwidth=1,

    font=dict(
        size=12
    ),

    text=insight_text
)
# ---------------------------------------------------
# Highlight Highest Rated Year
# ---------------------------------------------------

fig.add_scatter(
    x=[highest_year],
    y=[highest_rating],

    mode='markers+text',

    text=[f"Highest Rated Year ({highest_year})"],

    textposition="top center",

    marker=dict(
        size=14,
        color="red"
    ),

    name="Highest Rated Year"
)

# ---------------------------------------------------
# Show Chart
# ---------------------------------------------------

fig.show()
```

#### 3.7 Are newer movies/series receiving higher or lower ratings compared to older movies?

```python
# @title
import pandas as pd
import plotly.graph_objects as go

# ---------------------------------------------------
# Data Preparation
# ---------------------------------------------------

# Convert ratings to numeric
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

# Keep valid ratings only
filtered_df = df[
    (df['rating'] <= 10) &
    (df['start_year'] >= 1900) &
    (df['start_year'] <= 2020)
]

# ---------------------------------------------------
# Separate Old vs New Eras
# ---------------------------------------------------

old_movies = filtered_df[
    (filtered_df['start_year'] >= 1900) &
    (filtered_df['start_year'] <= 1940)
]

new_movies = filtered_df[
    (filtered_df['start_year'] >= 1980) &
    (filtered_df['start_year'] <= 2020)
]

# Average ratings per year
old_trend = (
    old_movies
    .groupby('start_year')['rating']
    .mean()
    .reset_index(name='avg_rating')
)

new_trend = (
    new_movies
    .groupby('start_year')['rating']
    .mean()
    .reset_index(name='avg_rating')
)

# ---------------------------------------------------
# Calculate Insights
# ---------------------------------------------------

old_avg = round(old_trend['avg_rating'].mean(), 2)
new_avg = round(new_trend['avg_rating'].mean(), 2)

comparison = (
    "higher"
    if new_avg > old_avg
    else "lower"
)

# ---------------------------------------------------
# Create Visualization
# ---------------------------------------------------

fig = go.Figure()

# Old Movies Line
fig.add_trace(
    go.Scatter(
        x=old_trend['start_year'],
        y=old_trend['avg_rating'],
        mode='lines+markers',
        name='1900-1940 (Classic Era)',
        line=dict(color='#F5C518', width=3),
        marker=dict(size=6)
    )
)

# New Movies Line
fig.add_trace(
    go.Scatter(
        x=new_trend['start_year'],
        y=new_trend['avg_rating'],
        mode='lines+markers',
        name='1980-2020 (Modern Era)',
        line=dict(color='cyan', width=3),
        marker=dict(size=6)
    )
)

# ---------------------------------------------------
# Layout Styling
# ---------------------------------------------------

fig.update_layout(
    template='plotly_dark',

    title="<b>IMDb</b> | Classic vs Modern Movie Ratings",

    title_font=dict(
        size=24,
        color="#F5C518"
    ),

    xaxis_title="Release Year",
    yaxis_title="Average IMDb Rating",

    hovermode="x unified",

    margin=dict(r=350),

    legend=dict(
        x=0.02,
        y=0.98
    )
)

# ---------------------------------------------------
# Insight Annotation
# ---------------------------------------------------

insight_text = (
    f"<b>Key Insights</b><br>"
    f"• Classic Era Avg Rating: {old_avg}<br>"
    f"• Modern Era Avg Rating: {new_avg}<br>"
    f"• Modern movies generally receive <b>{comparison}</b> ratings<br>"
    f"• Comparison focuses on distinct historical periods"
)

fig.add_annotation(
    xref="paper",
    yref="paper",

    x=1.18,
    y=0.5,

    showarrow=False,

    align="left",

    bgcolor="rgba(0,0,0,0.7)",

    bordercolor="#F5C518",
    borderwidth=1,

    font=dict(size=12),

    text=insight_text
)

# ---------------------------------------------------
# Show Chart
# ---------------------------------------------------

fig.show()
```

```python
# @title
import pandas as pd
import plotly.express as px

# ---------------------------------------------------
# Data Preparation
# ---------------------------------------------------

# Convert duration column to numeric
df['duration'] = pd.to_numeric(df['duration'], errors='coerce')

# Filter years between 1900–2020
filtered_df = df[
    (df['end_year'] >= 1900) &
    (df['end_year'] <= 2020)
]

# Average duration per year
duration_trend = (
    filtered_df
    .groupby('end_year')['duration']
    .mean()
    .reset_index(name='average_duration')
)

duration_trend.columns = ['Year', 'Average Duration']

# ---------------------------------------------------
# Key Insights
# ---------------------------------------------------

highest_row = duration_trend.loc[
    duration_trend['Average Duration'].idxmax()
]

lowest_row = duration_trend.loc[
    duration_trend['Average Duration'].idxmin()
]

highest_year = int(highest_row['Year'])
highest_duration = round(highest_row['Average Duration'], 1)

lowest_year = int(lowest_row['Year'])
lowest_duration = round(lowest_row['Average Duration'], 1)

overall_avg = round(
    duration_trend['Average Duration'].mean(),
    1
)

# ---------------------------------------------------
# Create Visualization
# ---------------------------------------------------

fig = px.line(
    duration_trend,
    x='Year',
    y='Average Duration',

    title="<b>IMDb</b> | Average Movie/Series Duration Across Years",

    markers=True
)

# ---------------------------------------------------
# IMDb-inspired Styling
# ---------------------------------------------------

fig.update_layout(
    template="plotly_dark",

    title_font=dict(
        size=24,
        color="#F5C518"
    ),

    xaxis_title="Year",
    yaxis_title="Average Duration (Minutes)",

    hovermode="x unified",

    margin=dict(
        r=420,
        b=80
    ),

    xaxis=dict(
        range=[1900, 2020]
    )
)

# IMDb yellow styling
fig.update_traces(
    line=dict(
        color="#F5C518",
        width=3
    ),

    marker=dict(
        color="#F5C518",
        size=6
    )
)

# ---------------------------------------------------
# Insight Annotation
# ---------------------------------------------------

insight_text = (
    f"<b>Key Insights</b><br>"
    f"• Longest avg duration: {highest_duration} mins ({highest_year})<br>"
    f"• Shortest avg duration: {lowest_duration} mins ({lowest_year})<br>"
    f"• Overall avg duration: {overall_avg} mins<br>"
    f"• Runtime patterns vary noticeably across decades"
)

fig.add_annotation(
    xref="paper",
    yref="paper",

    x=2,
    y=0.5,

    showarrow=False,

    align="left",

    bgcolor="rgba(0,0,0,0.7)",

    bordercolor="#F5C518",
    borderwidth=1,

    font=dict(size=12),

    text=insight_text
)

# ---------------------------------------------------
# Show Visualization
# ---------------------------------------------------

fig.show()
```
